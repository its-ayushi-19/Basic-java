```python
%%writefile linear_regression.cu
#include <stdio.h>
#include <stdlib.h>
#include <math.h>
#include <omp.h>
#include <cuda_runtime.h>

#define CUDA_CHECK(call) do {                                      \
    cudaError_t err = (call);                                      \
    if (err != cudaSuccess) {                                      \
        fprintf(stderr, "CUDA error at %s:%d: %s\n",               \
                     __FILE__, __LINE__, cudaGetErrorString(err)); \
        exit(EXIT_FAILURE);                                        \
    }                                                               \
} while (0)

typedef struct {
    double w0, w1, w2;
} Weights;

double mse_cpu(const double* x1, const double* x2, const double* y, int N, Weights w) {
    double sum = 0.0;
    for (int i = 0; i < N; ++i) {
        double pred = w.w0 + w.w1 * x1[i] + w.w2 * x2[i];
        double e = pred - y[i];
        sum += e * e;
    }
    return sum / (2.0 * N);
}

void train_sequential(const double* x1, const double* x2, const double* y,
                      int N, int iterations, double lr, Weights* w) {
    for (int it = 0; it < iterations; ++it) {
        double g0 = 0.0, g1 = 0.0, g2 = 0.0;
        for (int i = 0; i < N; ++i) {
            double e = w->w0 + w->w1 * x1[i] + w->w2 * x2[i] - y[i];
            g0 += e;
            g1 += e * x1[i];
            g2 += e * x2[i];
        }
        double invN = 1.0 / N;
        w->w0 -= lr * g0 * invN;
        w->w1 -= lr * g1 * invN;
        w->w2 -= lr * g2 * invN;
    }
}

void train_openmp(const double* x1, const double* x2, const double* y,
                  int N, int iterations, double lr, Weights* w) {
    for (int it = 0; it < iterations; ++it) {
        double g0 = 0.0, g1 = 0.0, g2 = 0.0;
        #pragma omp parallel for reduction(+:g0,g1,g2) schedule(static)
        for (int i = 0; i < N; ++i) {
            double e = w->w0 + w->w1 * x1[i] + w->w2 * x2[i] - y[i];
            g0 += e;
            g1 += e * x1[i];
            g2 += e * x2[i];
        }
        double invN = 1.0 / N;
        w->w0 -= lr * g0 * invN;
        w->w1 -= lr * g1 * invN;
        w->w2 -= lr * g2 * invN;
    }
}

__global__ void gradient_reduce_kernel(const double* x1, const double* x2,
                                       const double* y, int N,
                                       const double* weights,
                                       double* pg0, double* pg1, double* pg2) {
    extern __shared__ double shared[];
    double* s0 = shared;
    double* s1 = shared + blockDim.x;
    double* s2 = shared + 2 * blockDim.x;

    int i = blockIdx.x * blockDim.x + threadIdx.x;
    double a = 0.0, b = 0.0, c = 0.0;
    if (i < N) {
        double e = weights[0] + weights[1] * x1[i] + weights[2] * x2[i] - y[i];
        a = e;
        b = e * x1[i];
        c = e * x2[i];
    }

    s0[threadIdx.x] = a;
    s1[threadIdx.x] = b;
    s2[threadIdx.x] = c;
    __syncthreads();

    for (unsigned int stride = blockDim.x / 2; stride > 0; stride >>= 1) {
        if (threadIdx.x < stride) {
            s0[threadIdx.x] += s0[threadIdx.x + stride];
            s1[threadIdx.x] += s1[threadIdx.x + stride];
            s2[threadIdx.x] += s2[threadIdx.x + stride];
        }
        __syncthreads();
    }

    if (threadIdx.x == 0) {
        pg0[blockIdx.x] = s0[0];
        pg1[blockIdx.x] = s1[0];
        pg2[blockIdx.x] = s2[0];
    }
}

__global__ void update_weights_kernel(const double* pg0, const double* pg1,
                                      const double* pg2, int numBlocks, int N,
                                      double lr, double* weights) {
    if (blockIdx.x == 0 && threadIdx.x == 0) {
        double g0 = 0.0, g1 = 0.0, g2 = 0.0;
        for (int b = 0; b < numBlocks; ++b) {
            g0 += pg0[b];
            g1 += pg1[b];
            g2 += pg2[b];
        }
        double invN = 1.0 / N;
        weights[0] -= lr * g0 * invN;
        weights[1] -= lr * g1 * invN;
        weights[2] -= lr * g2 * invN;
    }
}

void train_cuda(const double* hx1, const double* hx2, const double* hy,
                int N, int iterations, double lr, Weights* w, double* kernel_ms) {
    int threads = 256;
    int blocks = (N + threads - 1) / threads;

    double *dx1 = NULL, *dx2 = NULL, *dy = NULL;
    double *dpg0 = NULL, *dpg1 = NULL, *dpg2 = NULL;
    double *dw = NULL;

    CUDA_CHECK(cudaMalloc((void**)&dx1, (size_t)N * sizeof(double)));
    CUDA_CHECK(cudaMalloc((void**)&dx2, (size_t)N * sizeof(double)));
    CUDA_CHECK(cudaMalloc((void**)&dy,  (size_t)N * sizeof(double)));
    CUDA_CHECK(cudaMalloc((void**)&dpg0, (size_t)blocks * sizeof(double)));
    CUDA_CHECK(cudaMalloc((void**)&dpg1, (size_t)blocks * sizeof(double)));
    CUDA_CHECK(cudaMalloc((void**)&dpg2, (size_t)blocks * sizeof(double)));
    CUDA_CHECK(cudaMalloc((void**)&dw, 3 * sizeof(double)));

    CUDA_CHECK(cudaMemcpy(dx1, hx1, (size_t)N * sizeof(double), cudaMemcpyHostToDevice));
    CUDA_CHECK(cudaMemcpy(dx2, hx2, (size_t)N * sizeof(double), cudaMemcpyHostToDevice));
    CUDA_CHECK(cudaMemcpy(dy,  hy,  (size_t)N * sizeof(double), cudaMemcpyHostToDevice));

    double init[3] = {w->w0, w->w1, w->w2};
    CUDA_CHECK(cudaMemcpy(dw, init, sizeof(init), cudaMemcpyHostToDevice));

    cudaEvent_t start, stop;
    CUDA_CHECK(cudaEventCreate(&start));
    CUDA_CHECK(cudaEventCreate(&stop));
    CUDA_CHECK(cudaEventRecord(start, 0));

    for (int it = 0; it < iterations; ++it) {
        gradient_reduce_kernel<<<blocks, threads, (size_t)3 * threads * sizeof(double)>>>(
            dx1, dx2, dy, N, dw, dpg0, dpg1, dpg2);
        CUDA_CHECK(cudaGetLastError());

        update_weights_kernel<<<1, 1>>>(dpg0, dpg1, dpg2, blocks, N, lr, dw);
        CUDA_CHECK(cudaGetLastError());
    }

    CUDA_CHECK(cudaEventRecord(stop, 0));
    CUDA_CHECK(cudaEventSynchronize(stop));
    float elapsed = 0.0f;
    CUDA_CHECK(cudaEventElapsedTime(&elapsed, start, stop));
    *kernel_ms = (double)elapsed;

    double finalW[3];
    CUDA_CHECK(cudaMemcpy(finalW, dw, sizeof(finalW), cudaMemcpyDeviceToHost));
    w->w0 = finalW[0];
    w->w1 = finalW[1];
    w->w2 = finalW[2];

    CUDA_CHECK(cudaEventDestroy(start));
    CUDA_CHECK(cudaEventDestroy(stop));
    CUDA_CHECK(cudaFree(dx1));
    CUDA_CHECK(cudaFree(dx2));
    CUDA_CHECK(cudaFree(dy));
    CUDA_CHECK(cudaFree(dpg0));
    CUDA_CHECK(cudaFree(dpg1));
    CUDA_CHECK(cudaFree(dpg2));
    CUDA_CHECK(cudaFree(dw));
}

int main() {
    int N = 5000000;          // 5 million data points
    int iterations = 500;     // Increased iterations for better convergence
    double lr = 0.05;

    // Force OpenMP to utilize multiple CPU cores (e.g., 4 threads) so it beats sequential cleanly
    omp_set_num_threads(4);

    double *x1 = (double*)malloc(N * sizeof(double));
    double *x2 = (double*)malloc(N * sizeof(double));
    double *y  = (double*)malloc(N * sizeof(double));

    if (!x1 || !x2 || !y) {
        fprintf(stderr, "Memory allocation failed!\n");
        return EXIT_FAILURE;
    }

    for (int i = 0; i < N; ++i) {
        x1[i] = (double)((i * 17LL) % 1000) / 1000.0;
        x2[i] = (double)((i * 29LL + 7) % 1000) / 1000.0;
        double noise = (double)((i * 13LL) % 101 - 50) / 10000.0;
        y[i] = 3.0 + 2.0 * x1[i] - 1.5 * x2[i] + noise;
    }

    Weights ws = {0.0, 0.0, 0.0};
    Weights wo = {0.0, 0.0, 0.0};
    Weights wg = {0.0, 0.0, 0.0};

    // 1. Sequential Timing
    double start_time = omp_get_wtime();
    train_sequential(x1, x2, y, N, iterations, lr, &ws);
    double seq_ms = (omp_get_wtime() - start_time) * 1000.0;

    // 2. OpenMP Timing (using multiple active threads)
    omp_set_dynamic(0);
    start_time = omp_get_wtime();
    train_openmp(x1, x2, y, N, iterations, lr, &wo);
    double omp_ms = (omp_get_wtime() - start_time) * 1000.0;

    // 3. CUDA GPU Timing
    double cuda_ms = 0.0;
    train_cuda(x1, x2, y, N, iterations, lr, &wg, &cuda_ms);

    double mse_s = mse_cpu(x1, x2, y, N, ws);
    double mse_o = mse_cpu(x1, x2, y, N, wo);
    double mse_g = mse_cpu(x1, x2, y, N, wg);

    printf("\nDatapoints: %d | Iterations: %d | Learning Rate: %.2f\n", N, iterations, lr);
    printf("OpenMP Active Threads: %d\n\n", omp_get_max_threads());
    printf("%-18s %14s %14s %14s\n", "Version", "Time (ms)", "Speedup", "MSE");
    printf("------------------------------------------------------------\n");
    printf("%-18s %14.4f %14.4f %14.4f\n", "Sequential", seq_ms, 1.0, mse_s);
    printf("%-18s %14.4f %14.4f %14.4f\n", "OpenMP", omp_ms, seq_ms / omp_ms, mse_o);
    printf("%-18s %14.4f %14.4f %14.4f\n", "CUDA GPU", cuda_ms, seq_ms / cuda_ms, mse_g);

    free(x1);
    free(x2);
    free(y);
    return 0;
}
```
